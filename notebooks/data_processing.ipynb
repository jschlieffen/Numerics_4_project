```python
import pandas as pd
from datetime import datetime
```

```python
df = pd.read_csv("../data/zib/Infection_data/Our_World_In_Data/owid-covid-data.csv", index_col="date", parse_dates=["date"])
```

```python
df_deu = df[df["location"] == "Germany"]
idx = df_deu["new_cases"].notna()
df_deu = df_deu[idx].loc[:, ["new_cases", "new_cases_smoothed", "total_cases"]]
```

```python
start_date = datetime(2020, 3, 1)
end_date = datetime(2020, 6, 1)

df_deu.loc[start_date:end_date].plot(y="new_cases", title="New COVID-19 Cases in Germany")
```
