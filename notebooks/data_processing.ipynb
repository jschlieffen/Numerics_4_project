```python
import pandas as pd
from datetime import datetime
```

```python
start_date = datetime(2020, 3, 12)
end_date = datetime(2020, 6, 26)

infection_data = pd.read_csv("../data/zib/Infection_data/Our_World_In_Data/daily-new-estimated-infections-of-covid-19.csv", index_col="Day", parse_dates=["Day"])
```

```python
infection_data = infection_data[infection_data["Code"] == "JPN"]
infection_data = infection_data.loc[start_date:end_date]
infection_data
```

```python
label = "Daily new estimated COVID-19 infections (LSHTM, Median estimate)"
infection_data.loc[:, [label]].plot()
```
